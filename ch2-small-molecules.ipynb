### The Basics with Water

In this **series of notebooks** we will explore how to describe a molecule to *PSi4* using the example of water. We will then use various methods to **optimize the structure** and determine its electronic **energy**. 

[Structure of Water](./2A-water-structure/ch2A-water.ipynb)
: In this notebook we will **describe** water using a set of **Cartesian coordinates** (the classic $xyz$ system) and a set of **internal coordinates** (bond distances and angles relative to a starting point). We can also obtain starting structures from databases or *Python* tools like `rdkit`.

Level of Effort
: No pain, no gain. The low-level methods used above give **adequate**, but not high-quality results. In this notebook, we will perform geometry optimizations and **compare** different levels of **computational effort** using algorithms that include more information about the electrons and larger basis sets that describe the atomic orbitals more accurately. We will **compare** the results to experimental observations.

Vibrations
: Rock and roll! Psi4 can calculate the **bond vibrations** and they can be **visualized** within the notebook. We can also **compare** the level of effort with experimental results.

Molecular Orbitals
: We can get the **energy levels** for the molecular orbitals. We can also **visualize** the MOs in the notebook. It is also possible to pick apart the MOs and isolate the contributions from "**local**" bonding and lone pair orbitals. These match with the $sp^3$ hybridization model that most organic chemists use when imagining the **electronic structure** of water.

Surfaces
: We can compute a surface that closely matches the van der Waals surface of the molecule and then calculate properties that can be mapped onto that surface as color. We will visualize the polar nature of water that way.

Bond Dissociation
: We can pull the O–H bond apart to follow the bond dissociation. Will it follow the Morse potential or the harmonic potential? Can we get a value for the bond dissociation energy? How do different levels of effort affect the results?

Acidic and Basic Water
: We will calculate the energy for $\ce{H3O+ -> H2O + H+}$ and $\ce{H2O -> HO- + H+}$.

Hydrogen Bonding
: All these calculations are in gas phase. Two water molecules will want to hydrogen bond. We can calculate the structure of a water dimer and a small water cluster. Does the structure align better with the canonical MOs or the local MOs? What is the energy for H-bonding in gas phase?
