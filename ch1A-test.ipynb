## Your First Calculation

Let us try a real calculation. The code below will **calculate the energy** of a water molecule when you execute it. The next chapter will describe in detail how to use *Psi4*



````python
import psi4 

# set output file to water.out, and overwrite if it already exists
psi4.set_output_file("water.out", False)   # False = overwrite; True = append
psi4.set_memory('500 MB') # set memory to 500 MB

# create a water molecule in Z-matrix format
zmatrix = """
O
H 1 0.96
H 1 0.96 2 104.5
"""

# create a molecule object from the Z-matrix
psi4.geometry(zmatrix)

# Calculate the HF energy of water using the cc-pVDZ basis set
energy = psi4.energy('hf/cc-pvdz')
print(f"Energy: {energy} Hartrees")
````



After a few seconds you should see an energy of about `-76.02` Hartrees. It works!
